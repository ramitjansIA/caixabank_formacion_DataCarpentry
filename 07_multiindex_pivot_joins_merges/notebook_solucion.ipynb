## 1. Preparación

```python
import time
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

SEED = 42
rng = np.random.default_rng(SEED)

try:
    import polars as pl
    POLARS_DISPONIBLE = True
    print("Polars:", pl.__version__)
except ImportError:
    POLARS_DISPONIBLE = False
    print("Polars no instalado. Ejecuta: pip install polars")

print("Pandas:", pd.__version__)
```

# 2. Crear tablas relacionadas

```python
n_clientes = 30_000
n_ventas = 300_000

clientes = pd.DataFrame({
    "cliente_id": np.arange(1, n_clientes + 1),
    "segmento": rng.choice(
        ["Particular", "PyME", "Enterprise"],
        n_clientes,
        p=[0.65, 0.27, 0.08]
    ),
    "region_cliente": rng.choice(
        ["Norte", "Sur", "Este", "Oeste", "Centro"],
        n_clientes
    )
})

ventas = pd.DataFrame({
    "venta_id": np.arange(1, n_ventas + 1),
    "cliente_id": rng.integers(1, n_clientes + 1501, n_ventas),
    "region": rng.choice(
        ["Norte", "Sur", "Este", "Oeste", "Centro"],
        n_ventas
    ),
    "canal": rng.choice(
        ["web", "tienda", "partner", "marketplace"],
        n_ventas
    ),
    "producto": rng.choice(
        [f"P{i:02d}" for i in range(1, 21)],
        n_ventas
    ),
    "unidades": rng.integers(1, 8, n_ventas),
    "precio": rng.uniform(10, 300, n_ventas),
    "fecha": pd.Timestamp("2025-01-01") + pd.to_timedelta(
        rng.integers(0, 365 * 24 * 60, n_ventas),
        unit="m"
    )
})

ventas["importe"] = ventas["unidades"] * ventas["precio"]

print("Clientes:", clientes.shape)
print("Ventas:", ventas.shape)
```

### Inconsistencia deliberada

`ventas` contiene algunos `cliente_id` que **no existen** en la tabla maestra `clientes`.

No los corregiremos todavía. Primero veremos cómo un join permite detectarlos.

# 3. Granularidad antes de unir

```python
print("¿cliente_id único en clientes?:", clientes["cliente_id"].is_unique)
print("¿venta_id único en ventas?:", ventas["venta_id"].is_unique)
print("¿cliente_id único en ventas?:", ventas["cliente_id"].is_unique)
```

### Interpretación

- `clientes`: una fila por cliente.
- `ventas`: una fila por operación.
- un cliente puede tener muchas ventas.

La relación esperada al añadir atributos de cliente a ventas es **muchos-a-uno (`m:1`)**.

# 9. `pivot()` — primero un caso válido

```python
tabla_simple = pd.DataFrame({
    "region": ["Norte", "Norte", "Sur", "Sur"],
    "canal": ["web", "tienda", "web", "tienda"],
    "valor": [100, 120, 90, 150]
})

tabla_simple.pivot(
    index="region",
    columns="canal",
    values="valor"
)
```

# 10. Provocar fallo de `pivot()` por duplicados

```python
tabla_duplicada = pd.DataFrame({
    "region": ["Norte", "Norte", "Norte", "Sur"],
    "canal": ["web", "web", "tienda", "web"],
    "valor": [100, 80, 120, 90]
})

tabla_duplicada
```

```python
try:
    tabla_duplicada.pivot(
        index="region",
        columns="canal",
        values="valor"
    )
except ValueError as error:
    print(type(error).__name__, "->", error)
```

### Diagnóstico

`pivot()` necesita una correspondencia única para cada combinación `index × columns`.

La combinación `Norte-web` aparece dos veces, por lo que Pandas no puede decidir qué valor colocar.

# 11. Corregir con `pivot_table()`

```python
tabla_pivot = tabla_duplicada.pivot_table(
    index="region",
    columns="canal",
    values="valor",
    aggfunc="sum",
    fill_value=0
)

tabla_pivot
```

### Interpretación

No hemos “arreglado” técnicamente el error sin más: hemos tomado una decisión analítica explícita.

Para duplicados `region-canal`, queremos **sumar** los valores.

# 12. Pivot table real

```python
pivot_real = ventas.pivot_table(
    index="region",
    columns="canal",
    values="importe",
    aggfunc="sum",
    fill_value=0
)

pivot_real
```

## 12.1 Heatmap

```python
fig, ax = plt.subplots(figsize=(8, 5))

im = ax.imshow(
    pivot_real.values,
    aspect="auto"
)

ax.set_xticks(
    range(len(pivot_real.columns)),
    pivot_real.columns
)
ax.set_yticks(
    range(len(pivot_real.index)),
    pivot_real.index
)

ax.set_title("Facturación: región × canal")

for i in range(pivot_real.shape[0]):
    for j in range(pivot_real.shape[1]):
        ax.text(
            j, i,
            f"{pivot_real.iloc[i, j] / 1e6:.1f}M",
            ha="center",
            va="center"
        )

plt.colorbar(im, ax=ax, label="Facturación")
plt.show()
```

# 13. Primer merge

```python
ventas_clientes = ventas.merge(
    clientes,
    on="cliente_id",
    how="left"
)

ventas_clientes.head()
```

## 13.1 ¿Hemos perdido filas?

```python
print("Ventas antes:", len(ventas))
print("Filas después:", len(ventas_clientes))
print(
    "Diferencia:",
    len(ventas_clientes) - len(ventas)
)
```

El número de filas se mantiene, pero eso no garantiza que el join sea correcto. Debemos comprobar también las correspondencias.

# 14. Auditar con `indicator=True`

```python
auditoria_join = ventas.merge(
    clientes,
    on="cliente_id",
    how="left",
    indicator=True
)

auditoria_join["_merge"].value_counts()
```

## 14.1 Visualizar correspondencias

```python
conteo_merge = (
    auditoria_join["_merge"]
    .value_counts()
    .reindex(
        ["both", "left_only", "right_only"],
        fill_value=0
    )
)

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(
    conteo_merge.index.astype(str),
    conteo_merge.values
)
ax.set(
    title="Auditoría del join ventas → clientes",
    xlabel="Estado",
    ylabel="Filas"
)
plt.show()
```

### Interpretación

Las filas `left_only` son ventas cuyo `cliente_id` no aparece en la tabla maestra.

El join ha convertido una inconsistencia referencial en algo medible.

# 15. Inspeccionar clientes huérfanos

```python
huerfanas = auditoria_join.loc[
    auditoria_join["_merge"] == "left_only",
    ["venta_id", "cliente_id", "importe"]
]

huerfanas.head()
```

```python
print(
    "Ventas huérfanas:",
    len(huerfanas)
)

print(
    "Clientes desconocidos:",
    huerfanas["cliente_id"].nunique()
)
```

# 16. Anti-join en Pandas

```python
anti_join_pd = (
    ventas.merge(
        clientes[["cliente_id"]],
        on="cliente_id",
        how="left",
        indicator=True
    )
    .loc[
        lambda df: df["_merge"] == "left_only"
    ]
    .drop(columns="_merge")
)

anti_join_pd.head()
```

Un anti-join devuelve las filas de la izquierda que **no encuentran correspondencia** en la derecha.

Es una herramienta de calidad de datos extremadamente útil.

# 17. Semi-join en Pandas

```python
ids_validos = clientes["cliente_id"]

semi_join_pd = ventas.loc[
    ventas["cliente_id"].isin(ids_validos)
]

print(semi_join_pd.shape)
```

Un semi-join conserva filas de la izquierda que sí tienen correspondencia, pero no añade columnas de la derecha.

# 18. Validar cardinalidad con `validate`

```python
merge_validado = ventas.merge(
    clientes,
    on="cliente_id",
    how="left",
    validate="m:1"
)

print("Merge m:1 validado.")
```

`validate="m:1"` convierte una expectativa conceptual en una comprobación ejecutable.

# 19. Provocar una violación de cardinalidad

```python
clientes_duplicados = pd.concat(
    [
        clientes,
        clientes.iloc[[0]]
    ],
    ignore_index=True
)

print(
    clientes_duplicados[
        clientes_duplicados["cliente_id"] == 1
    ]
)
```

```python
try:
    ventas.merge(
        clientes_duplicados,
        on="cliente_id",
        how="left",
        validate="m:1"
    )
except pd.errors.MergeError as error:
    print(type(error).__name__, "->", error)
```

### Diagnóstico

La tabla derecha ya no contiene una fila única por `cliente_id`.

Sin `validate`, el join podría ejecutarse y multiplicar filas silenciosamente.

# 20. Ver la explosión de filas sin `validate`

```python
ventas_cliente_1 = ventas.loc[
    ventas["cliente_id"] == 1
]

merge_explosivo = ventas_cliente_1.merge(
    clientes_duplicados,
    on="cliente_id",
    how="left"
)

print("Antes:", len(ventas_cliente_1))
print("Después:", len(merge_explosivo))
```

Una duplicación en la dimensión puede multiplicar cada fila relacionada. En datasets grandes, esto puede inflar métricas y memoria sin producir una excepción.

# 21. Corregir duplicados de clave

```python
duplicados_cliente = clientes_duplicados.loc[
    clientes_duplicados["cliente_id"].duplicated(
        keep=False
    )
]

duplicados_cliente
```

```python
clientes_corregidos = (
    clientes_duplicados
    .drop_duplicates(
        subset="cliente_id",
        keep="first"
    )
)

print(
    "Clave única:",
    clientes_corregidos["cliente_id"].is_unique
)
```

En un caso real no debemos aplicar `drop_duplicates()` automáticamente: primero hay que decidir **por qué existen duplicados y cuál es la regla correcta de resolución**.

# 22. Joins por varias claves

```python
tarifas = (
    ventas[["region", "canal"]]
    .drop_duplicates()
    .assign(
        coste_servicio=lambda df: rng.uniform(
            1, 8, len(df)
        )
    )
)

tarifas.head()
```

```python
ventas_tarifas = ventas.merge(
    tarifas,
    on=["region", "canal"],
    how="left",
    validate="m:1"
)

ventas_tarifas[[
    "region",
    "canal",
    "importe",
    "coste_servicio"
]].head()
```

# 23. Sufijos cuando existen columnas con el mismo nombre

```python
clientes_con_region = clientes.rename(
    columns={
        "region_cliente": "region"
    }
)

ejemplo_sufijos = ventas.head(5).merge(
    clientes_con_region,
    on="cliente_id",
    how="left",
    suffixes=("_venta", "_cliente")
)

ejemplo_sufijos[[
    "cliente_id",
    "region_venta",
    "region_cliente"
]]
```

Los sufijos hacen explícito el origen de variables homónimas. Renombrar correctamente es preferible a trabajar con `_x` y `_y` sin significado.

# 24. Detectar inconsistencias entre dos atributos

```python
comparacion_region = ventas.merge(
    clientes,
    on="cliente_id",
    how="inner",
    validate="m:1"
)

comparacion_region["region_coincide"] = (
    comparacion_region["region"]
    == comparacion_region["region_cliente"]
)

comparacion_region["region_coincide"].value_counts(normalize=True)
```

Este ejemplo ilustra que un join también puede utilizarse para **validar coherencia entre fuentes**, no solo para añadir columnas.

# 25. Joins temporales: problema

Supongamos que cada producto cambia de tarifa a lo largo del tiempo y queremos asignar a cada venta la última tarifa conocida **anterior o igual** a la fecha de venta.

```python
fechas_tarifa = pd.date_range(
    "2025-01-01",
    periods=12,
    freq="MS"
)

tarifas_tiempo = pd.DataFrame([
    {
        "producto": producto,
        "fecha_tarifa": fecha,
        "coste_base": float(rng.uniform(5, 120))
    }
    for producto in [f"P{i:02d}" for i in range(1, 21)]
    for fecha in fechas_tarifa
])

tarifas_tiempo.head()
```

# 26. `merge_asof()`

```python
ventas_asof = (
    ventas[[
        "venta_id",
        "producto",
        "fecha",
        "importe"
    ]]
    .sort_values("fecha")
)

tarifas_asof = (
    tarifas_tiempo
    .sort_values("fecha_tarifa")
)

resultado_asof = pd.merge_asof(
    ventas_asof,
    tarifas_asof,
    left_on="fecha",
    right_on="fecha_tarifa",
    by="producto",
    direction="backward"
)

resultado_asof.head()
```

### Interpretación

`merge_asof()` no busca igualdad exacta de fecha. Busca la observación temporal más adecuada según la dirección especificada.

Este patrón aparece en:

- precios históricos;
- sensores;
- mercados financieros;
- estados de dispositivos;
- configuraciones vigentes.

# 27. Validar que no usamos tarifas futuras

```python
violaciones_temporales = (
    resultado_asof["fecha_tarifa"]
    > resultado_asof["fecha"]
).sum()

print(
    "Tarifas futuras asignadas:",
    violaciones_temporales
)
```

# 28. Preparar Polars

```python
if POLARS_DISPONIBLE:
    ventas_pl = pl.from_pandas(ventas)
    clientes_pl = pl.from_pandas(clientes)
    tarifas_pl = pl.from_pandas(tarifas)
    tarifas_tiempo_pl = pl.from_pandas(tarifas_tiempo)

    print(ventas_pl.head())
else:
    print("Instalar Polars para continuar la parte comparativa.")
```

## Polars y MultiIndex

Polars no utiliza un índice de filas como Pandas.

Las dimensiones se mantienen como columnas explícitas. Por tanto, muchos flujos basados en:

- `set_index`;
- `reset_index`;
- `xs`;
- `stack/unstack`;

se replantean mediante columnas, `group_by`, `pivot`, `unpivot`, `select` y expresiones.

# 29. GroupBy multidimensional sin MultiIndex

```python
if POLARS_DISPONIBLE:
    resumen_pl = (
        ventas_pl
        .group_by(
            ["region", "canal"]
        )
        .agg(
            pl.col("importe")
            .sum()
            .alias("facturacion"),

            pl.len()
            .alias("operaciones")
        )
        .sort(
            ["region", "canal"]
        )
    )

    print(resumen_pl.head(10))
```

# 30. Pivot en Polars

```python
if POLARS_DISPONIBLE:
    pivot_pl = (
        resumen_pl
        .pivot(
            on="canal",
            index="region",
            values="facturacion",
            aggregate_function="sum"
        )
        .sort("region")
    )

    print(pivot_pl)
```

En Polars las dimensiones siguen siendo columnas; no aparece un MultiIndex como resultado del `group_by`.

# 31. Join left en Polars

```python
if POLARS_DISPONIBLE:
    ventas_clientes_pl = (
        ventas_pl
        .join(
            clientes_pl,
            on="cliente_id",
            how="left"
        )
    )

    print(ventas_clientes_pl.head())
    print("Filas:", ventas_clientes_pl.height)
```

# 32. Anti-join nativo en Polars

```python
if POLARS_DISPONIBLE:
    anti_pl = (
        ventas_pl
        .join(
            clientes_pl.select("cliente_id"),
            on="cliente_id",
            how="anti"
        )
    )

    print(anti_pl.head())
    print("Ventas huérfanas:", anti_pl.height)
```

### Ventaja expresiva

En Pandas construimos el anti-join mediante `merge(..., indicator=True)` y filtrado.

Polars dispone directamente de `how="anti"`.

# 33. Semi-join nativo en Polars

```python
if POLARS_DISPONIBLE:
    semi_pl = (
        ventas_pl
        .join(
            clientes_pl.select("cliente_id"),
            on="cliente_id",
            how="semi"
        )
    )

    print(semi_pl.head())
    print("Filas:", semi_pl.height)
```

# 34. Validar anti-join Pandas vs Polars

```python
if POLARS_DISPONIBLE:
    ids_pd = set(
        anti_join_pd["venta_id"].tolist()
    )

    ids_pl = set(
        anti_pl["venta_id"].to_list()
    )

    print(
        "Mismas ventas huérfanas:",
        ids_pd == ids_pl
    )
```

# 35. Join por varias claves en Polars

```python
if POLARS_DISPONIBLE:
    ventas_tarifas_pl = (
        ventas_pl
        .join(
            tarifas_pl,
            on=["region", "canal"],
            how="left"
        )
    )

    print(
        ventas_tarifas_pl.select(
            "region",
            "canal",
            "importe",
            "coste_servicio"
        ).head()
    )
```

# 36. Join temporal `join_asof()` en Polars

```python
if POLARS_DISPONIBLE:
    ventas_asof_pl = (
        ventas_pl
        .select(
            "venta_id",
            "producto",
            "fecha",
            "importe"
        )
        .sort("fecha")
    )

    tarifas_asof_pl = (
        tarifas_tiempo_pl
        .sort("fecha_tarifa")
    )

    resultado_asof_pl = (
        ventas_asof_pl
        .join_asof(
            tarifas_asof_pl,
            left_on="fecha",
            right_on="fecha_tarifa",
            by="producto",
            strategy="backward"
        )
    )

    print(resultado_asof_pl.head())
```

# 37. Validar join temporal

```python
if POLARS_DISPONIBLE:
    violaciones_pl = (
        resultado_asof_pl
        .filter(
            pl.col("fecha_tarifa")
            > pl.col("fecha")
        )
        .height
    )

    print(
        "Tarifas futuras asignadas:",
        violaciones_pl
    )
```

# 38. Join con validación de cardinalidad en Polars

```python
if POLARS_DISPONIBLE:
    try:
        _ = (
            ventas_pl
            .join(
                clientes_pl,
                on="cliente_id",
                how="left",
                validate="m:1"
            )
        )
        print("Cardinalidad m:1 validada en Polars.")
    except Exception as error:
        print(type(error).__name__, "->", error)
```

La validación de cardinalidad debe formar parte de los joins críticos siempre que conozcamos la relación esperada.

# 39. Benchmark de join Pandas vs Polars

```python
def medir(funcion, repeticiones=5):
    tiempos = []

    for _ in range(repeticiones):
        inicio = time.perf_counter()
        funcion()
        tiempos.append(
            time.perf_counter() - inicio
        )

    return np.array(tiempos)


def join_pandas():
    return ventas.merge(
        clientes,
        on="cliente_id",
        how="left",
        validate="m:1"
    )


def join_polars():
    if not POLARS_DISPONIBLE:
        return None

    return ventas_pl.join(
        clientes_pl,
        on="cliente_id",
        how="left",
        validate="m:1"
    )


t_pd = medir(join_pandas)

if POLARS_DISPONIBLE:
    t_pl = medir(join_polars)
else:
    t_pl = np.array([np.nan])

benchmark_join = pd.DataFrame({
    "motor": ["Pandas", "Polars"],
    "media_ms": [
        t_pd.mean() * 1000,
        t_pl.mean() * 1000
    ],
    "std_ms": [
        t_pd.std() * 1000,
        t_pl.std() * 1000
    ]
})

benchmark_join
```

## 39.1 Visualizar benchmark

```python
if POLARS_DISPONIBLE:
    fig, ax = plt.subplots(figsize=(6, 4))

    ax.bar(
        benchmark_join["motor"],
        benchmark_join["media_ms"],
        yerr=benchmark_join["std_ms"],
        capsize=4
    )

    ax.set(
        title="Join m:1 — Pandas vs Polars",
        ylabel="Tiempo medio (ms)"
    )

    plt.show()
```

### Interpretación

No debemos convertir un benchmark aislado en una regla universal. El rendimiento depende del tamaño, tipos, cardinalidad, hardware y pipeline completo.

## Ejercicio 4 — Cardinalidad

Construye ejemplos 1:1, 1:m, m:1 y m:m. Usa validate en cada uno y explica cuántas filas produce la multiplicidad de claves.


```python
izq_unica_practica = pd.DataFrame({"clave": [1, 2], "valor_izq": [10, 20]})
der_unica_practica = pd.DataFrame({"clave": [1, 2], "valor_der": [100, 200]})
izq_multiple_practica = pd.DataFrame({"clave": [1, 1, 2], "valor_izq": [10, 11, 20]})
der_multiple_practica = pd.DataFrame({"clave": [1, 1, 2], "valor_der": [100, 101, 200]})
casos_cardinalidad_practica = {
    "1:1": (izq_unica_practica, der_unica_practica, 2),
    "1:m": (izq_unica_practica, der_multiple_practica, 3),
    "m:1": (izq_multiple_practica, der_unica_practica, 3),
    "m:m": (izq_multiple_practica, der_multiple_practica, 5),
}
for cardinalidad, (izq, der, esperado) in casos_cardinalidad_practica.items():
    resultado = izq.merge(der, on="clave", validate=cardinalidad)
    assert len(resultado) == esperado
    print(cardinalidad, "filas:", len(resultado))
```

### Solución razonada

Las cardinalidades indican unicidad de claves a cada lado. Para una clave con dos filas a izquierda y dos a derecha, el join genera cuatro pares. validate="m:m" permite esa relación; no impone una restricción de unicidad.

## Ejercicio 12 — Asof

Asigna a cada observación el último estado anterior o igual. Incluye una observación previa a cualquier estado y conserva su falta de correspondencia.


```python
observaciones_tiempo_practica = pd.DataFrame({"observacion_id": [1, 2, 3, 4], "producto": ["A"] * 4,
    "fecha": pd.to_datetime(["2025-01-01", "2025-01-05", "2025-01-10", "2025-01-15"])})
estados_tiempo_practica = pd.DataFrame({"producto": ["A", "A"], "fecha_estado": pd.to_datetime(["2025-01-05", "2025-01-12"]), "estado": ["inicial", "nuevo"]})
asof_practica = pd.merge_asof(observaciones_tiempo_practica.sort_values("fecha"), estados_tiempo_practica.sort_values("fecha_estado"),
    left_on="fecha", right_on="fecha_estado", by="producto", direction="backward")
assert pd.isna(asof_practica.loc[0, "estado"])
assert asof_practica.loc[1:, "estado"].tolist() == ["inicial", "inicial", "nuevo"]
assert len(asof_practica) == len(observaciones_tiempo_practica)
print(asof_practica)
```

## Ejercicio 13 — Validación temporal

Valida ausencia de estados futuros y que el estado asignado sea el más reciente disponible. Compara el ejemplo temporal con Polars.

### Solución razonada

## Ejercicio 15 — Benchmark

Compara left join y anti-join entre motores con cinco repeticiones, entradas iguales y conversiones fuera de la medición. Valida filas, columnas, valores y nulos antes de medir.


```python
def medir_joins_practica(metodos, repeticiones=5):
    for funcion in metodos.values():
        funcion()
    medidas = []
    for repeticion in range(repeticiones):
        orden = list(metodos.items())
        if repeticion % 2:
            orden.reverse()
        for nombre, funcion in orden:
            inicio = time.perf_counter()
            resultado = funcion()
            medidas.append({"repeticion": repeticion + 1, "metodo": nombre, "ms": (time.perf_counter() - inicio) * 1000})
    tabla = pd.DataFrame(medidas)
    resumen = tabla.groupby("metodo")["ms"].agg(media="mean", desviacion="std").reset_index()
    return tabla, resumen

def graficar_joins_practica(resumen, titulo):
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.bar(resumen["metodo"], resumen["media"], yerr=resumen["desviacion"], capsize=5)
    ax.set(title=titulo, ylabel="ms (media ± desviación estándar)")
    ax.tick_params(axis="x", rotation=20)
    plt.tight_layout()
    plt.show()
if POLARS_DISPONIBLE:
    def left_pd_practica():
        return ventas.merge(dimension_limpia_practica, on="cliente_id", how="left", validate="m:1")
    def left_pl_practica():
        return hechos_pl_practica.join(dimension_pl_practica, on="cliente_id", how="left", validate="m:1")
    def anti_pd_practica():
        return ventas.loc[~ventas["cliente_id"].isin(dimension_limpia_practica["cliente_id"])]
    def anti_pl_funcion_practica():
        return hechos_pl_practica.join(dimension_pl_practica, on="cliente_id", how="anti")

    comparacion_left_pd = left_pd_practica().sort_values("venta_id").reset_index(drop=True)
    comparacion_left_pl = left_pl_practica().sort("venta_id").to_pandas()[comparacion_left_pd.columns]
    pd.testing.assert_frame_equal(comparacion_left_pd, comparacion_left_pl)
    pd.testing.assert_frame_equal(anti_pd_practica().sort_values("venta_id").reset_index(drop=True), anti_pl_funcion_practica().sort("venta_id").to_pandas())
    tiempos_left_practica, resumen_left_practica = medir_joins_practica({"Pandas left": left_pd_practica, "Polars left": left_pl_practica})
    tiempos_anti_practica, resumen_anti_practica = medir_joins_practica({"Pandas anti": anti_pd_practica, "Polars anti": anti_pl_funcion_practica})
    graficar_joins_practica(resumen_left_practica, "Left join: cinco repeticiones")
    graficar_joins_practica(resumen_anti_practica, "Anti-join: cinco repeticiones")
    print(resumen_left_practica)
    print(resumen_anti_practica)
else:
    print("Pendiente: instalar polars y pyarrow para ejecutar esta parte.")
```

### Solución razonada

Los tiempos describen cálculo sobre tablas cargadas. Las conversiones para validar ocurren fuera de la medición. La dispersión representa variación entre repeticiones; no determina que un motor sea universalmente mejor. El left conserva nulos para clientes huérfanos.

Crea tres tablas con al menos 300.000 hechos: ventas, clientes e historial de tarifas. Introduce y diagnostica huérfanos y una duplicación exacta. Define una política de corrección; valida m:1 y vigencia temporal. Implementa ambos motores, semi/anti, pivot y benchmark.


```python
# Tres tablas: hechos de venta, dimensión cliente y estados de tarifa por producto/fecha.
hechos_reto_joins = ventas.copy(deep=True)
assert len(hechos_reto_joins) >= 300_000
copia_hechos_reto = hechos_reto_joins.copy(deep=True)
dimension_reto_joins = pd.concat([clientes, clientes.iloc[[0]]], ignore_index=True)
assert dimension_reto_joins.duplicated("cliente_id").any()
dimension_reto_joins = dimension_reto_joins.drop_duplicates().copy()
assert dimension_reto_joins["cliente_id"].is_unique
productos_reto = sorted(hechos_reto_joins["producto"].unique())
estados_reto_joins = pd.DataFrame([
    {"producto": producto, "fecha_tarifa": pd.Timestamp(fecha), "tarifa": 10.0 + i + periodo}
    for i, producto in enumerate(productos_reto)
    for periodo, fecha in enumerate(["2024-12-01", "2025-04-01", "2025-08-01"])])
assert not estados_reto_joins.duplicated(["producto", "fecha_tarifa"]).any()
anti_reto_pd = hechos_reto_joins.loc[~hechos_reto_joins["cliente_id"].isin(dimension_reto_joins["cliente_id"])].copy()
semi_reto_pd = hechos_reto_joins.loc[hechos_reto_joins["cliente_id"].isin(dimension_reto_joins["cliente_id"])].copy()
assert len(anti_reto_pd) > 0
assert len(anti_reto_pd) + len(semi_reto_pd) == len(hechos_reto_joins)
# Política: excluir huérfanos del análisis y conservarlos como tabla de incidencias.
enriquecido_reto_pd = semi_reto_pd.merge(dimension_reto_joins, on="cliente_id", how="left", validate="m:1")
temporal_reto_pd = pd.merge_asof(enriquecido_reto_pd.sort_values("fecha"), estados_reto_joins.sort_values("fecha_tarifa"),
    left_on="fecha", right_on="fecha_tarifa", by="producto", direction="backward")
assert len(temporal_reto_pd) == len(semi_reto_pd)
assert temporal_reto_pd["venta_id"].is_unique
assert temporal_reto_pd["fecha_tarifa"].notna().all()
assert temporal_reto_pd["fecha_tarifa"].le(temporal_reto_pd["fecha"]).all()
# Independiente del asof: comprobar el último periodo de vigencia esperado.
fecha_esperada_reto = pd.Series(pd.Timestamp("2024-12-01"), index=temporal_reto_pd.index)
fecha_esperada_reto.loc[temporal_reto_pd["fecha"].ge("2025-04-01")] = pd.Timestamp("2025-04-01")
fecha_esperada_reto.loc[temporal_reto_pd["fecha"].ge("2025-08-01")] = pd.Timestamp("2025-08-01")
np.testing.assert_array_equal(temporal_reto_pd["fecha_tarifa"].to_numpy(), fecha_esperada_reto.to_numpy())
resumen_reto_joins = temporal_reto_pd.groupby(["segmento", "canal"], as_index=False).agg(facturacion=("importe", "sum"), operaciones=("venta_id", "size"))
assert resumen_reto_joins["operaciones"].sum() == len(semi_reto_pd)
pivot_reto_joins = resumen_reto_joins.pivot(index="segmento", columns="canal", values="facturacion")
pd.testing.assert_frame_equal(hechos_reto_joins, copia_hechos_reto)
if POLARS_DISPONIBLE:
    hechos_reto_pl = pl.from_pandas(hechos_reto_joins)
    dimension_reto_pl = pl.from_pandas(dimension_reto_joins)
    estados_reto_pl = pl.from_pandas(estados_reto_joins)
    anti_reto_pl = hechos_reto_pl.join(dimension_reto_pl, on="cliente_id", how="anti")
    semi_reto_pl = hechos_reto_pl.join(dimension_reto_pl, on="cliente_id", how="semi")
    np.testing.assert_array_equal(np.sort(anti_reto_pd["venta_id"]), np.sort(anti_reto_pl["venta_id"].to_numpy()))
    np.testing.assert_array_equal(np.sort(semi_reto_pd["venta_id"]), np.sort(semi_reto_pl["venta_id"].to_numpy()))
    temporal_reto_pl = semi_reto_pl.join(dimension_reto_pl, on="cliente_id", how="left", validate="m:1").sort("fecha").join_asof(
        estados_reto_pl.sort("fecha_tarifa"), left_on="fecha", right_on="fecha_tarifa", by="producto", strategy="backward")
    tabla_pd = temporal_reto_pd.sort_values("venta_id").reset_index(drop=True)
    tabla_pl = temporal_reto_pl.sort("venta_id").to_pandas()[tabla_pd.columns]
    pd.testing.assert_frame_equal(tabla_pd, tabla_pl)
    resumen_reto_pl = temporal_reto_pl.group_by(["segmento", "canal"]).agg(
        pl.col("importe").sum().alias("facturacion"), pl.len().alias("operaciones")).sort(["segmento", "canal"]).to_pandas()
    resumen_pd_ordenado = resumen_reto_joins.sort_values(["segmento", "canal"]).reset_index(drop=True)
    assert resumen_pd_ordenado[["segmento", "canal"]].equals(resumen_reto_pl[["segmento", "canal"]])
    np.testing.assert_allclose(resumen_pd_ordenado["facturacion"], resumen_reto_pl["facturacion"], rtol=1e-10)
    np.testing.assert_array_equal(resumen_pd_ordenado["operaciones"], resumen_reto_pl["operaciones"])
    tiempos_reto_joins, resumen_tiempos_reto_joins = medir_joins_practica({
        "Pandas m:1": lambda: semi_reto_pd.merge(dimension_reto_joins, on="cliente_id", how="left", validate="m:1"),
        "Polars m:1": lambda: semi_reto_pl.join(dimension_reto_pl, on="cliente_id", how="left", validate="m:1")})
    graficar_joins_practica(resumen_tiempos_reto_joins, "Reto: enriquecimiento m:1")
else:
    print("Parte Polars pendiente de instalar dependencias.")
ax = pivot_reto_joins.plot(kind="bar", figsize=(9, 4))
ax.set(title="Reto: facturación por segmento y canal", xlabel="Segmento", ylabel="Facturación")
plt.tight_layout()
plt.show()
print("Granularidad: una fila por venta; una por cliente; una por producto y fecha de tarifa.")
print("Huérfanos conservados como incidencias:", len(anti_reto_pd))
print("Reto verificado: cardinalidad, cobertura, vigencia y equivalencia.")
```

### Solución razonada

**Política de calidad:** eliminar solo duplicaciones idénticas; apartar huérfanos como incidencias y analizarlos sin inventar atributos. La tarifa se asigna desde la última vigencia disponible y no cambia el importe observado de la venta.

## Respuestas orientativas

1. MultiIndex representa claves jerárquicas y facilita selecciones por niveles, a costa de una semántica adicional de indexación y cambios de forma.
2. pivot exige una observación por combinación de fila y columna; varias observaciones por celda producen ambigüedad.
3. pivot_table decide cómo agregar esas observaciones; aggfunc debe responder a la pregunta y a la granularidad.
4. Cardinalidad describe si la clave es única o múltiple en cada lado del join.
5. Un m:m genera pares y puede repetir hechos e inflar métricas si la multiplicidad no es intencional.
6. validate comprueba la relación esperada y rechaza claves incompatibles antes de aceptar el join.
7. indicator identifica coincidencias y filas exclusivas de cada lado según el tipo de join.
8. Semi conserva filas izquierdas con coincidencia; anti conserva las que no tienen coincidencia. No añaden atributos derechos.
9. Puede conservarse el total de filas y seguir habiendo huérfanos, asignaciones erróneas o valores nulos. Se deben comprobar claves y cobertura.
10. Cuando se necesita el último estado disponible anterior o igual a cada observación, con historial temporal y una política explícita.
11. Introduce información no disponible al momento analizado y puede causar leakage o conclusiones inválidas.
12. Las claves permanecen como columnas explícitas y las operaciones usan esas columnas; se alinea por identificadores cuando hace falta.
13. Permiten separar hechos válidos e incidencias sin incorporar atributos ni multiplicar filas.
14. Lazy permite optimizar el plan completo, incluidos filtros y proyecciones; la ventaja de tiempo debe medirse.
15. Exigir el mismo contrato de claves, nulos, cardinalidad, cobertura, vigencia, columnas y valores, con tolerancias numéricas justificadas y coste de conversión conocido.
